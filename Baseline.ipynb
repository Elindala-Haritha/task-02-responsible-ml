```python
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
```

```python
print("Scikit-learn is working!")
```

```python
X = df[
    [
        "tenure_months",
        "support_tickets",
        "monthly_spend_inr",
        "last_login_days",
        "plan_type"
    ]
]

y = df["churned"]

print("Features shape:", X.shape)
print("Target shape:", y.shape)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=42,
    stratify=y
)

print("Training rows:", len(X_train))
print("Testing rows:", len(X_test))
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        (
            "categorical",
            OneHotEncoder(handle_unknown="ignore"),
            ["plan_type"]
        )
    ],
    remainder="passthrough"
)

model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("classifier", LogisticRegression(max_iter=1000))
    ]
)
```

```python
model.fit(X_train, y_train)

print("Model training completed!")
```

```python
y_pred = model.predict(X_test)

print("Actual values:", y_test.tolist())
print("Predicted values:", y_pred.tolist())
```

```python
accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred, zero_division=0)
recall = recall_score(y_test, y_pred, zero_division=0)
f1 = f1_score(y_test, y_pred, zero_division=0)

print("Model Accuracy:", accuracy)
print("Model Precision:", precision)
print("Model Recall:", recall)
print("Model F1-score:", f1)
```

```python
baseline_accuracy = accuracy_score(y_test, [0] * len(y_test))

print("Baseline Accuracy:", baseline_accuracy)
print("Model Accuracy:", accuracy)

if accuracy > baseline_accuracy:
    print("The model performed better than the baseline.")
elif accuracy == baseline_accuracy:
    print("The model performed the same as the baseline.")
else:
    print("The model performed worse than the baseline.")
```
